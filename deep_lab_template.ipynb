```python
from torchvision.models import ResNet50_Weights
from torchvision.models.segmentation import deeplabv3_resnet50, DeepLabV3_ResNet50_Weights
import torch.nn as nn
```

```python
model = deeplabv3_resnet50(weights=DeepLabV3_ResNet50_Weights.DEFAULT, weights_backbone=ResNet50_Weights.DEFAULT)
print(model.classifier)
total = sum(p.numel() for p in model.parameters())
trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)

print("Total params:", total)
print("Trainable params:", trainable)
```

```python
num_classes = 19

model = deeplabv3_resnet50(weights=DeepLabV3_ResNet50_Weights.DEFAULT, weights_backbone=ResNet50_Weights.DEFAULT)
model.classifier[4] = nn.Conv2d(in_channels=256, out_channels=num_classes, kernel_size=1, stride=1)
```
